```python
bins = [0, 12, 24, 36, 48, 60, 72]
labels = ['0–12', '13–24', '25–36', '37–48', '49–60', '61+']

dados_telecomx['tenure_group'] = pd.cut(
    dados_telecomx['tenure'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
churn_tenure = (
    dados_telecomx
    .groupby(['tenure_group', 'Churn'])
    .size()
    .unstack()
)

churn_rate_tenure = (
    churn_tenure
    .div(churn_tenure.sum(axis=1), axis=0) * 100
)
```

```python
churn_contract = (
    dados_telecomx
    .groupby(['Contract', 'Churn'])
    .size()
    .unstack()
)
```

```python
churn_rate_contract = (
    churn_contract
    .div(churn_contract.sum(axis=1), axis=0) * 100
)
```

```python
order = ['Month-to-month', 'One year', 'Two year']
churn_rate_contract = churn_rate_contract.loc[order]
```

```python
bins = [0, 30, 60, 90, 120, 150]
labels = ['0-30', '31-60', '61-90', '91-120', '121+']

dados_telecomx['charges_group'] = pd.cut(
    dados_telecomx['MonthlyCharges'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
churn_charges = (
    dados_telecomx
    .groupby(['charges_group', 'Churn'])
    .size()
    .unstack()
)

churn_rate_charges = (
    churn_charges
    .div(churn_charges.sum(axis=1), axis=0) * 100
)
```

```python
payment_method_churn = dados_telecomx.groupby('Churn')['PaymentMethod'].value_counts().unstack()
print(payment_method_churn)
```

```python
churn_rate_payment = (
    payment_method_churn
    .div(payment_method_churn.sum(axis=1), axis=0) * 100
)
```

```python
servicos_adicionais = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]
```

```python
resultados = []

for servico in servicos_adicionais:
    churn_servico = (
        dados_telecomx
        .groupby([servico, 'Churn'])
        .size()
        .unstack()
    )

    churn_rate = (
        churn_servico
        .div(churn_servico.sum(axis=1), axis=0) * 100
    )

    churn_rate = churn_rate.reset_index()

    churn_rate['Servico'] = servico
    churn_rate.rename(columns={servico: 'StatusServico'}, inplace=True)

    resultados.append(churn_rate)
```

```python
servicos = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

for servico in servicos:
    print(f'\n{servico}')
    print(
        dados_telecomx
        .groupby(servico)['Churn']
        .value_counts(normalize=True)
        .mul(100)
        .round(1)
    )
```

```python
servicos = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

tabelas = []

for servico in servicos:
    temp = (
        dados_telecomx
        .groupby(servico)['Churn']
        .value_counts(normalize=True)
        .mul(100)
        .round(1)
        .unstack()
        .reset_index()
    )

    temp['Servico'] = servico
    temp.rename(columns={servico: 'StatusServico'}, inplace=True)
    tabelas.append(temp)

tabela_final = pd.concat(tabelas, ignore_index=True)

tabela_final
```

```python
servicos = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

dados_plot = []

for servico in servicos:
    temp = (
        dados_telecomx
        .groupby(servico)['Churn']
        .value_counts(normalize=True)
        .mul(100)
        .round(1)
        .unstack()
        .reset_index()
    )

    temp = temp[temp[servico].isin(['Yes', 'No'])]
    temp['Servico'] = servico
    temp.rename(columns={servico: 'Status'}, inplace=True)

    dados_plot.append(temp[['Servico', 'Status', 'Yes']])

dados_plot = pd.concat(dados_plot, ignore_index=True)
```

```python
churn_gender = (
    dados_telecomx
    .groupby('gender')['Churn']
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
    .unstack()
)
```

#📄Relatorio Final


**Base: Dados de clientes de telecomunicações**

## 1. Introdução

O churn representa a evasão de clientes e é um dos principais desafios estratégicos em empresas de telecomunicações. Este relatório tem como objetivo identificar **padrões associados ao churn**, analisando variáveis contratuais, financeiras, comportamentais, demográficas e relacionadas a serviços adicionais, além de propor **ações práticas para mitigar a evasão**.

## 2. Churn e Tempo de Contrato (Tenure)

### 🔍 Análise

A análise da taxa de churn por tempo de contrato evidencia que clientes com **menor tempo de permanência** apresentam taxas significativamente mais altas de evasão. Observa-se uma **redução progressiva do churn** à medida que o tempo de relacionamento com a empresa aumenta.

### 🧠 Interpretação

Esse comportamento indica que o churn está fortemente concentrado nos **primeiros meses de contrato**, período no qual o cliente ainda está avaliando o valor do serviço e é mais sensível a frustrações iniciais.

### 🔁 Possíveis ações para redução

* Estruturação de um **onboarding ativo** nos primeiros meses
* Monitoramento de clientes com até 12 meses de contrato
* Ações preventivas de retenção no início da jornada

## 3. Churn e Tipo de Contrato

### 🔍 Análise

Clientes com contratos do tipo **month-to-month** apresentam as maiores taxas de churn, enquanto contratos de **maior duração** (one year e two year) demonstram taxas significativamente menores.

### 🧠 Interpretação

Contratos mensais reduzem a barreira de saída e facilitam a evasão, enquanto contratos de longo prazo refletem maior comprometimento e confiança do cliente na empresa.

### 🔁 Possíveis ações para redução

* Incentivos financeiros para migração para contratos anuais
* Benefícios exclusivos para clientes com contratos longos
* Comunicação clara do valor agregado desses planos

## 4. Churn e Cobrança Mensal (Monthly Charges)

### 🔍 Análise

A análise da média, e do gráfico de densidade (KDE) indica que clientes que realizaram churn tendem a apresentar **cobranças mensais mais elevadas**, com maior concentração em faixas de preço mais altas.

### 🧠 Interpretação

O churn está associado à **percepção de custo-benefício**. Clientes que pagam mais possuem expectativas mais altas e, quando o valor percebido não acompanha o preço, a evasão se torna mais provável.

### 🔁 Possíveis ações para redução

* Revisão de planos de alto valor
* Reforço na comunicação dos benefícios dos planos mais caros
* Inclusão de serviços adicionais para aumentar o valor percebido

## 5. Churn e Método de Pagamento

### 🔍 Análise

Clientes que utilizam **métodos de pagamento manuais**, especialmente *electronic check*, apresentam taxas de churn mais elevadas. Já métodos automáticos, como cartão de crédito e débito automático, estão associados a menores taxas de evasão.

### 🧠 Interpretação

Métodos manuais aumentam a fricção no relacionamento com o cliente, enquanto métodos automáticos indicam maior engajamento e facilitam a retenção.

### 🔁 Possíveis ações para redução

* Incentivo à adesão a pagamentos automáticos
* Benefícios para clientes que utilizam débito automático
* Simplificação do processo de cobrança

## 6. Churn e Serviços Adicionais

### 🔍 Análise

A comparação entre clientes que **possuem** e **não possuem** serviços adicionais demonstra que a ausência desses serviços está associada a **taxas de churn significativamente maiores**. O impacto é mais expressivo em serviços como **TechSupport** e **OnlineSecurity**.

### 🧠 Interpretação

Serviços adicionais aumentam o valor percebido e fortalecem o vínculo do cliente com a empresa. Clientes sem esses serviços tendem a perceber menos benefícios e evadir com maior facilidade.

### 🔁 Possíveis ações para redução

* Criação de pacotes (bundles) com serviços adicionais
* Ofertas de teste gratuito de serviços estratégicos
* Comunicação clara do valor desses serviços

## 7. Churn e Gênero

### 🔍 Análise

A análise da taxa de churn por gênero indica que **não há diferenças expressivas** entre os grupos. As taxas de churn entre gêneros são bastante próximas, sugerindo impacto limitado dessa variável na evasão.

### 🧠 Interpretação

O gênero, isoladamente, **não é um fator determinante** para o churn. Esse resultado reforça que a evasão está mais associada a fatores contratuais, financeiros e de valor percebido do que a características demográficas.

### 🔁 Implicações estratégicas

* Estratégias de retenção **não devem ser segmentadas prioritariamente por gênero**
* O foco deve permanecer em variáveis comportamentais e contratuais

## 8. Conclusão Geral

A análise integrada evidencia que o churn **não ocorre de forma aleatória**, mas está fortemente associado a:

* baixo tempo de contrato
* contratos mensais
* cobranças mensais elevadas
* métodos de pagamento manuais
* ausência de serviços adicionais

Variáveis demográficas, como gênero, apresentam impacto reduzido na evasão.

### 🎯 Estratégia integrada de retenção

Uma estratégia eficaz para redução do churn deve:

* atuar preventivamente nos primeiros meses de contrato
* incentivar contratos de longo prazo
* reduzir fricções operacionais
* aumentar o valor percebido por meio de serviços adicionais
